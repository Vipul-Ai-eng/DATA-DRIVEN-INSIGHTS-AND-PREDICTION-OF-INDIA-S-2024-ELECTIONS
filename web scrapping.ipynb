```python
from bs4 import BeautifulSoup
import pandas as pd

# Step 1: Load the saved HTML file
with open(r"C:\Users\vipul\Downloads\Candidates with declared criminal cases_Lok Sabha 2024 Election page 2.html", "r", encoding="utf-8") as f:
    html = f.read()

# Step 2: Parse with BeautifulSoup
soup = BeautifulSoup(html, "lxml")

# Step 3: Find the main data table
table = soup.find("table", {"class": "w3-table"})

# Print table structure to debug
if table:
    # Extract the actual column headers from the table
    headers = [th.get_text(strip=True) for th in table.find("tr").find_all("th")]
    print(f"Found {len(headers)} columns: {headers}")
    
    rows = []
    for tr in table.find_all("tr")[1:]:  # skip header
        cols = [td.get_text(strip=True) for td in tr.find_all("td")]
        if cols:
            rows.append(cols)
            
    # Step 4: Create DataFrame with the actual headers from the table
    if not headers or len(headers) != len(rows[0]):
        headers = [f"Column_{i}" for i in range(len(rows[0]))]
    
    df = pd.DataFrame(rows, columns=headers)
    
    # Step 5: Save as CSV
    df.to_csv("myneta_page2.csv", index=False, encoding="utf-8")
    print("Data extracted and saved to myneta_page2.csv")
else:
    print("Table not found. Check the HTML structure or class name.")
```
